# Data Fetching Overview

This notebook demonstrates how to fetch and compress datasets for different financial instruments using the `fetch_data` function. The data types covered include **bond**, **fixing**, and **repo**. Each data type is retrieved using three different compression methods:

1. **compress_by_column_all_instruments** – Compresses data by a specific column across all instruments.
2. **compress_by_instrument_all_columns** – Compresses all columns for a specific instrument.
3. **compress_full_dataset** – Compresses the entire dataset without filtering.


## Bond Data

```
df_bond_compress_by_column_all_instruments = fetch_data(
    start_date=start_date,
    end_date=end_date,
    data_type='bond',
    method='compress_by_column_all_instruments',
    columns_to_analyze=['Price B']
)

df_bond_compress_by_instrument_all_columns = fetch_data(
    start_date=start_date,
    end_date=end_date,
    data_type='bond',
    method='compress_by_instrument_all_columns',
    instrument_name='DS0725'
)

df_bond_compress_full_dataset = fetch_data(
    start_date=start_date,
    end_date=end_date,
    data_type='bond',
    method='compress_full_dataset'
)
```

```
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

def plot_bond_return_indices(df):
    """
    Plots normalized simple return indices for bond instruments based on 'PriceB' columns.
    - Normalizes prices to start from 1
    - Removes '_PriceB' from legend labels
    - Adds a title indicating it's based on 'PriceB'
    - Places the legend at the bottom of the plot
    """
    # Ensure index is datetime for proper plotting
    if not np.issubdtype(df.index.dtype, np.datetime64):
        df.index = pd.to_datetime(df.index)

    # Convert all columns to float if they are object type
    for col in df.columns:
        if df[col].dtype == object:
            df[col] = df[col].str.replace(',', '').astype(float)

    # Normalize prices to start from 1
    df_normalized = df / df.iloc[0]

    # Prepare labels without '_PriceB'
    labels = [col.replace('_PriceB', '') for col in df.columns]

    # Plotting
    plt.figure(figsize=(16, 9))
    for i, col in enumerate(df.columns):
        plt.plot(df_normalized.index, df_normalized[col], label=labels[i])

    plt.title("Normalized Return Indices Based on 'PriceB'", fontsize=16)
    plt.xlabel("Date")
    plt.ylabel("Normalized Price (Start = 1)")
    plt.grid(True)
    plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=4, fontsize=9)
    plt.tight_layout()
    plt.show()

plot_bond_return_indices(df)
```
